# Introduction

In this session, we will use a simple **single-degree-of-freedom (SDOF) structural system** to introduce **object-oriented programming (OOP)** in Python.

The objective is not to develop a complete dynamic analysis tool, but to use a familiar structural-dynamics problem to understand how Python can be used to organise engineering calculations into **functions, classes, objects, and methods**.

## Single-degree-of-freedom system

A damped SDOF system can be represented by a mass-spring-damper system:

$$
m\ddot{x}(t) + c\dot{x}(t) + kx(t) = F(t)
$$

where:

- $m$ is the mass [kg]
- $c$ is the damping coefficient [N·s/m]
- $k$ is the stiffness [N/m]
- $x(t)$ is the displacement [m]
- $\dot{x}(t)$ is the velocity [m/s]
- $\ddot{x}(t)$ is the acceleration [m/s²]
- $F(t)$ is the applied external force [N]

For the purpose of this workshop, we will focus on some fundamental properties of the system rather than solving the complete equation of motion.

## Natural frequency

The undamped natural angular frequency is given by:

$$
\omega_n = \sqrt{\frac{k}{m}}
$$

where $\omega_n$ is in rad/s.

The natural frequency in Hertz is:

$$
f_n = \frac{\omega_n}{2\pi}
$$

Therefore:

$$
f_n = \frac{1}{2\pi}\sqrt{\frac{k}{m}}
$$

## Critical damping and damping ratio

The critical damping coefficient is:

$$
c_c = 2\sqrt{km}
$$

The damping ratio is defined as:

$$
\zeta = \frac{c}{c_c}
$$

or:

$$
\zeta = \frac{c}{2\sqrt{km}}
$$

The damping ratio provides a convenient way of describing the amount of damping in the system.

For example:

- $\zeta = 0$: undamped system
- $0 < \zeta < 1$: underdamped system
- $\zeta = 1$: critically damped system
- $\zeta > 1$: overdamped system

## Why use this problem for Python?

For a single structural system, we can simply define:

```python
mass = 5
stiffness = 200
damping = 0.5

# 1. Functions <a class="anchor" id="func"></a>

We can proceed by writing simple functions first

````python
# Import dependencies
import numpy as np
import matplotlib.pyplot as plt
````

````python
# System properties
m = 5       # mass [kg]
k = 200     # stiffness [N/m]
c = 100     # damping coefficient [N·s/m]
````

````python
# Write a simple code for f_n, c_critical and zeta

f_n = np.sqrt(k / m) / (2 * np.pi)

c_critical = 2 * np.sqrt(k * m)

zeta = c / c_critical

print(f"Natural frequency: {f_n:.2f} Hz")
print(f"Critical damping: {c_critical:.2f} N·s/m")
print(f"Damping ratio: {zeta:.3f}")
````



We have just calculated several properties of our SDOF system directly in the notebook.

However, imagine that we want to perform the same calculations for many different structural systems.

Instead of repeatedly writing the same equations, we can encapsulate each calculation in a **function**.

A function allows us to:

- give a piece of code a meaningful name,
- provide inputs,
- perform a calculation, and
- returnto a function.

````python
def natural_frequency(mass, stiffness):
    return np.sqrt(stiffness / mass) / (2 * np.pi)
````

````python
fn = natural_frequency(m, k)
print(f"Natural Frequency: {fn:.2f} Hz")
````

````python
def critical_damping(mass, stiffness):
    return 2 * np.sqrt(mass * stiffness)
````

````python
c_critical = critical_damping(m, k)

print(f"Critical damping: {c_critical:.2f} N·s/m")
````

````python
def damping_ratio(mass, stiffness, damping):
    return damping / (2 * np.sqrt(mass * stiffness))
````

````python
zeta = damping_ratio(m, k, c)

print(f"Damping ratio: {zeta:.3f}")
````

````python
natural_frequency(m, k)

critical_damping(m, k)

damping_ratio(m, k, c)
````

# 3. Classes and Objects

So far, we have used functions to perform calculations for our SDOF system.

For example:

```python
f_n = natural_frequency(m, k)
zeta = damping_ratio(m, k, c)

However, as the number of systems increases, this approach quickly becomes difficult to manage.

## What is a Class?

A **class** is a **blueprint or template for creating objects**.

For our structural-dynamics problem, we can define a class called `SDOFSystem` that describes what an SDOF system should contain.

The class can define both:

- **Properties** — the data or characteristics of the system, such as mass, stiffness, and damping.
- **Methods** — the operations or behaviours that can be performed on the system, such as calculating the natural frequency or damping ratio.

In simple terms:

> **A class defines what an object contains and what it can do.**

## What is an Object?

An **object** is a **specific instance of a class**.

The class `SDOFSystem` describes the general structure of an SDOF system, while `system_1`, `system_2`, and `system_3` can represent **specific structural systems** created from that class.

Each object:

- has its **own values** for properties such as mass, stiffness, and damping;
- follows the **same structure** defined by the class; and
- can use the **methods** defined by the class.

In simple terms:

> **A class is the blueprint, while an object is a specific instance built from that blueprint.**

For example:

```text
             SDOFSystem
           (Class / Blueprint)
                  │
        ┌─────────┼─────────┐
        ↓         ↓         ↓
    system_1  system_2  system_3
     (Object)  (Object)  (Object)

Each object has its own values, but all objects follow the same structure defined by the class.

## Why Use Classes?

Classes allow us to **group data and the operations performed on that data together**.

Instead of managing:

**Mass + Stiffness + Damping**

as separate variables:

```text
mass
stiffness
damping

The **object** contains the **properties of the system** and the **methods that operate on those properties**.

This makes our code:

- 📦 **Organised** — related information is kept together.
- ♻️ **Reusable** — we can create many objects from the same class.
- 👀 **Readable** — the code structure reflects the engineering problem.
- 🔧 **Extendable** — we can add new properties and calculations later.

> 💡 **Key idea:** An object combines **data (properties)** and **behaviour (methods)** into a single entity.

## Creating Our First Class

Let's create a class called `SDOFSystem` to represent our structural system.

The class will contain the three basic properties we have already defined:

- **Mass** $m$
- **Stiffness** $k$
- **Damping coefficient** $c$

We use the `__init__()` method to initialise these properties when a new object is created.

The keyword `self` refers to the **objecf.damping = damping

````python
class SDOFSystem:
    def __init__(self, mass, stiffness, damping):
        self.mass = mass
        self.stiffness = stiffness
        self.damping = damping
````

````python
system = SDOFSystem(m, k, c)
````

## Adding Methods

Our `SDOFSystem` class currently contains the **properties** of the structural system:

- Mass
- Stiffness
- Damping coefficient

However, a structural system is not only described by its properties. We also want to perform calculations using those properties.

For example, we may want to calculate:

- **Natural frequency**
- **Critical damping**
- **Damping ratio**

These calculations can be implemented as **methods** inside our class.

A method is a function that belongs to a class and can operate on the properties of an object.

For example, instead of calculating the natural frequency using:

```python
natural_frequency(m, k)

we can let the object calculate its own natural frequency:
```python
system.natural_frequency()

The method can access the object's properties using **`self`**.

> 💡 **Key idea:** A **method** is a function defined inside a class that operates on the **object's data**.

````python
class SDOFSystem:
    
    def __init__(self, mass, stiffness, damping):
        self.mass = mass
        self.stiffness = stiffness
        self.damping = damping

    def natural_frequency(self):
        return np.sqrt(self.stiffness / self.mass) / (2 * np.pi)

    def critical_damping(self):
        return 2 * np.sqrt(self.mass * self.stiffness)

    def damping_ratio(self):
        return self.damping / self.critical_damping()
        
````

````python
system = SDOFSystem(m, k, c)

print(f"Natural frequency: {system.natural_frequency():.2f} Hz")
print(f"Critical damping: {system.critical_damping():.2f} N·s/m")
print(f"Damping ratio: {system.damping_ratio():.3f}")
````

Here you can already notice the difference, instead of managing each property separatly, we just manage each object.

````python
damping_ratio(m, k, c)
````

````python
system.damping_ratio()
````

# 5. Creating and Analyzing Multiple Objects

Now that we have created our `SDOFSystem` class, we can use it to create **multiple structural systems**.

Each object will represent a different SDOF system with its own:

- **Mass**
- **Stiffness**
- **Damping coefficient**

Although the objects are created from the same class, each object can have different property values.

For example:

```text
SDOFSystem
     │
     ├── system_1 → m = 5 kg,   k = 200 N/m,  c = 0.5 N·s/m
     ├── system_2 → m = 10 kg,  k = 500 N/m,  c = 1.0 N·s/m
     └── system_3 → m = 20 kg,  k = 1000 N/m, c = 2.0 N·s/m

````python
system_1 = SDOFSystem(
    mass=5,
    stiffness=200,
    damping=0.5
)

system_2 = SDOFSystem(
    mass=10,
    stiffness=500,
    damping=1.0
)

system_3 = SDOFSystem(
    mass=20,
    stiffness=1000,
    damping=2.0
)
````

````python
systems = [system_1, system_2, system_3]

for system in systems:
    print(f"Natural frequency: {system.natural_frequency():.2f} Hz")
    print(f"Damping ratio: {system.damping_ratio():.3f}")
    print()
````

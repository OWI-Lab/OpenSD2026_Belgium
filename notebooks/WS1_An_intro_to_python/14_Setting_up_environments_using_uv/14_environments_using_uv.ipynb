# 1. Why do we need Python environments?

Before using `uv`, let's understand the problem it is designed to solve.

When working on engineering or research projects, we usually rely on many Python packages:

- `numpy` for numerical calculations
- `pandas` for data processing
- `matplotlib` for visualization
- `scipy` for scientific computing
- `scikit-learn` for machine learning

As projects become larger, managing these packages and their versions becomes increasingly important.

# The problem: Python projects have dependencies

Imagine that you are working on **two different projects**.

| | 🔧 **Structural Dynamics** | 🤖 **Machine Learning** |

| **SciPy** | 1.11 | — |
| **Matplotlib** | 3.8 | — |
| **pandas** | — | 2.x |
| **scikit-learn** | — | 1.6 |

You may encounter errors such as:

| Problem | What it means |
|:---|:---|

| **Version conflicts** | Packages require incompatible versions |
| **DLL errors** | Binary dependencies are incompatible |
| **Unexpected behaviour** | Code runs, but produces unexpected results |

# The solution: isolated environments

A Python environment provides an **isolated space** for a project.

Instead of putting every project's dependencies into one global Python installation,  
we create a **separate environment for each project**.

---

## ❌ One global environment

Everything is installed into the same Python environment:

## Try and run this code with the new environment activated

````python
import numpy as np
import matplotlib.pyplot as plt


def plot_sin_cos(duration=10, n_points=1000):
    """Plot sine and cosine signals over time."""
    
    t = np.linspace(0, duration, n_points)
    
    plt.plot(t, np.sin(t), label="sin(t)")
    plt.plot(t, np.cos(t), label="cos(t)")
    
    plt.xlabel("Time [s]")
    plt.ylabel("Amplitude")
    plt.legend()
    plt.grid()
    plt.show()


plot_sin_cos()
````

## 2. A few Git and GitHub concepts

Git is a **version control system**: it keeps track of every change ever made to a project, who made it, and why. GitHub is a website that hosts Git repositories and adds collaboration features (issues, pull requests, code review, etc.) on top of Git.

Here are the concepts you need for today:

- **Repository ("repo")** — a project's folder, together with its *entire history* of changes. The repository we will work with is:
  [OWI-Lab/OpenSD2026_Brussels_intro_git](https://github.com/OWI-Lab/OpenSD2026_Brussels_intro_git)
- **Commit** — a saved snapshot of changes, with a short message describing *what* changed and *why*. Commits are the building blocks of a repository's history; think of them as "save points" you can always return to.
- **Branch** — an independent line of development. The `main` branch usually holds the official, working version of the project. When you want to make a change, you create your *own* branch (e.g. `add-jane-doe`) so that your work-in-progress doesn't disturb `main` until it is ready and reviewed.
- **Pull request (PR)** — a request to merge the changes from your branch into another branch (usually `main`). A PR is where teammates review your code, leave comments, and discuss the change *before* it becomes part of the project.
- **Merge** — the act of combining the changes from one branch into another. Once a pull request is approved, it gets merged, and its commits become part of the target branch's history.
- **Merge conflict** — what happens when two branches change the *same lines* of the *same file* in different ways. Git cannot automatically decide which change should "win", so a human has to resolve the conflict by hand.

A typical open-source workflow therefore looks like this:


1. ‼️ Clone/pull the repository to get the latest main branch
2. Create a new branch for your change
3. Edit files and commit your changes
4. Push the branch and open a pull request
5. Discuss, review, resolve any conflicts
6. Merge the pull request into main

**Important** Especially the first step is important, always make sure your (local) copy is up to date. If you start working on old version, you risk a lot of unnecessary work!

We will now go through this workflow ourselves.

## 3. Get the repository onto your machine

There are two easy ways to work on the repository. Pick whichever you are more comfortable with:

**Option A — GitHub Desktop**

1. Install [GitHub Desktop](https://desktop.github.com/) if you don't already have it.
2. In the repository on GitHub, click the green **"Code"** button, then **"Open with GitHub Desktop"** (or copy the URL and use *File → Clone repository* in GitHub Desktop).
3. Choose a local folder and click **Clone**.

**Option B — GitHub Codespaces**

1. On the repository page, click the green **"Code"** button, open the **"Codespaces"** tab, and click **"Create codespace on main"**.
2. This spins up a ready-to-use VS Code environment in your browser with the repository already cloned — no local installation needed.

Either way, you now have your own local (or cloud) copy of the `main` branch, in sync with GitHub.

## 4. Create a branch and add yourself to the list

1. Create a new branch off `main`. Give it a short, descriptive name, e.g. `add-<your-name>`.
   - GitHub Desktop: use the **"Current Branch"** dropdown → **"New Branch"**.
   - Codespaces / VS Code: use the branch indicator in the bottom-left corner, or run `git checkout -b add-<your-name>` in the terminal.
2. Open `src/py_opensdbrussels2026/data/participants.csv` and add a new row with your `family_name`, `first_name`, and `country` (matching the existing columns).
3. Save the file.
4. Commit your change:
   - GitHub Desktop: write a short commit summary (e.g. "Add <your name> to participants list") and click **Commit to add-\<your-name\>**.
   - Terminal: `git add src/py_opensdbrussels2026/data/participants.csv && git commit -m "Add <your name> to participants list"`
5. Push your branch to GitHub (**Publish branch** in GitHub Desktop, or `git push -u origin add-<your-name>` in the terminal).
6. Open a **pull request** from your branch into `main` (GitHub Desktop will offer a button for this, or use the **"Compare & pull request"** button that appears on the GitHub repository page).

## 5. Merge conflicts happen — let's resolve one

Everyone in the room is adding a row to the *same file*, at roughly the *same location* (the end of the CSV). The first pull request that gets merged into `main` will go through cleanly. Every pull request opened *after* that will now conflict with `main`, because both versions changed the same lines: Git doesn't know how to combine them automatically.

When this happens, GitHub will show **"This branch has conflicts that must be resolved"** on your pull request page. To fix it:

1. Update your local branch with the latest `main` (pull/merge `main` into your branch, e.g. `git fetch origin && git merge origin/main`, or use the equivalent buttons in GitHub Desktop).
2. Git will mark the conflicting section in `participants.csv` with conflict markers:

   ```
   <<<<<<< HEAD
   Doe,Jane,Belgium
   =======
   Smith,John,Belgium
   >>>>>>> main
   ```

3. Open the file and edit it by hand: remove the `<<<<<<<`, `=======`, and `>>>>>>>` markers, and keep **both** rows (yours *and* the other person's), each on its own line.
4. Save the file, then stage and commit the resolved version (`git add ...` + `git commit`, or use the *Resolve conflicts* / commit UI in GitHub Desktop).
5. Push the branch again. The conflict on your pull request should now be gone, and it can be merged.

This is completely normal — merge conflicts are a routine part of collaborating on shared files, not a sign that something went wrong!

## 6. Pick up the fix: upgrade the package with `uv`

Once your pull request (and everyone else's) has been merged into `main`, a new version of `py-opensdbrussels2026` gets published with the updated participant list. To pick up the change in *this* workshop repository, upgrade the dependency with [`uv`](https://docs.astral.sh/uv/):

````python
!uv add py-opensdbrussels2026 --upgrade-package py-opensdbrussels2026
````

This tells `uv` to re-resolve and reinstall `py-opensdbrussels2026` at its latest available version, updating `pyproject.toml` and `uv.lock` accordingly.

‼️ **Important** Now restart the notebook kernel (so Python re-imports the updated package) and check again whether you show up in the attendee list.

````python
from py_opensdbrussels2026 import attendee_was_present

attendee_was_present("YourFirstName", "YourFamilyName")
````

If this returns `True`, congratulations — you have just completed the full open-source contribution loop: **clone → branch → commit → pull request → resolve conflicts → merge → upgrade with `uv`**. This is, in essence, how the vast majority of real-world Python packages (including the ones you rely on every day, like `numpy` or `pandas`) are built and maintained.
